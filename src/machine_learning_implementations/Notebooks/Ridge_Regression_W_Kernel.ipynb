# Formal Loss fucntion (primal problem)
$$\mathcal{L(\bf{w})} = \frac{1}{2}\|\mathcal{\bf{Xw-y}}\|_2^2+\frac{\lambda}{2}\|\mathcal{\bf{w}}\|_2^2$$

# Formal object function
$$\min_{w}{\mathcal{L(\bf{w})} = \frac{1}{2}\|\mathcal{\bf{Xw-y}}\|_2^2+\frac{\lambda}{2}\|\mathcal{\bf{w}}\|_2^2}$$

By taking the derivative of the loss function with respect to $\mathcal{w}$ we obtain following equation

$$\frac{\partial \mathcal{L}}{\partial \mathcal{w}}\mathcal{L(\bf{w})} = \frac{1}{2}\|\mathcal{\bf{Xw-y}}\|_2^2+\frac{\lambda}{2}\|\mathcal{\bf{w}}\|_2^2 = \bf{X}^{\top}[\bf{X}\mathcal{\bf{w}}-\bf{y}] + \lambda \bf{X} = 0$$

which further simplifies to 

$$\bf{X}^{\top}\bf{X}\mathcal{\bf{w}}-\bf{X}^{\top}\bf{y}+\lambda \bf{X} = 0$$

Now solving for $\mathcal{\bf{w}}$ and utlizing the push-through identity we get

$$\mathcal{\bf{w}} = \bf{X}^{\top}[\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]^{-1}\bf{y}$$

From the kernel trick we know that $\mathcal{\bf{w}} = \bf{X}^{\top}\bf{\alpha}$ therfore we know $\bf{\alpha}$ in this case is

$$\bf{\alpha} = [\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]^{-1}\bf{y}$$

Now we can obtain $\bf{\alpha}$ by saying now

$$[\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]\bf{\alpha} = \bf{y}$$

This is in terms of the derivative of the loss function so to back out the loss function w.r.t $\bf{\alpha}$ we do the following

1. Set the equation equal to zero like the formal derivative of the loss function was before
    - $[\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]\bf{\alpha} - \bf{y} = 0$
2. Take the integral w.r.t $\bf{\alpha}$
    - $\int{[\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]\bf{\alpha} - \bf{y}} = \frac{1}{2}\bf{\alpha}^{\top}[\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]\bf{\alpha} - \bf{\alpha}^{\top}\bf{y}$
3. Therfore the loss function is $\mathcal{L(\bf{\alpha})}$ is now
    - $\mathcal{L(\bf{\alpha})} = \frac{1}{2}\bf{\alpha}^{\top}[\bf{X}^{\top}\bf{X}+\lambda \bf{I}_N]\bf{\alpha} - \bf{\alpha}^{\top}\bf{y}$

The kernel K is then $\bf{K} = \bf{X}^{\top}\bf{X}$ so 

### Dual problem

$$\min_{\bf{\alpha}}\mathcal{L(\bf{\alpha})} = \frac{1}{2}\bf{\alpha}^{\top}[\bf{K}+\lambda \bf{I}_N]\bf{\alpha} - \bf{\alpha}^{\top}\bf{y}$$
